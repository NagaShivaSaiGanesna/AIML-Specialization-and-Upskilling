```python
import pandas as pd

# Creating a sample dataset
data = {
    'city': ['New York', 'London', 'Paris', 'Tokyo', 'New York', 'Paris'],
    'price': [200, 150, 320, 250, 180, 300]
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)
```

```python
# Calculate the mean price for each city and convert to a dictionary
mean_price = df.groupby('city')['price'].mean().to_dict()

print("\nMapping Dictionary:")
print(mean_price) 
# Output: {'London': 150.0, 'New York': 190.0, 'Paris': 310.0, 'Tokyo': 250.0}
```

```python
# Map the dictionary values to the original 'city' column
df['city_encoded'] = df['city'].map(mean_price)

# Select only the features needed for model training (dropping the raw text column)
final_training_data = df[['city_encoded', 'price']]

print("\nFinal Data for Model Training:\n", final_training_data)
```

```python
import seaborn as sns
import pandas as pd

# 1. Load the dataset
tips_df = sns.load_dataset('tips')

# 2. View the relevant columns
print(tips_df[['time', 'total_bill']].head())

# 3. Calculate the mean total_bill for each time category
time_mean_bill = tips_df.groupby('time')['total_bill'].mean().to_dict()
print("\nMean mapping:", time_mean_bill)

# 4. Map the means back to a new encoded column
tips_df['time_encoded'] = tips_df['time'].map(time_mean_bill)

# 5. Verify the transformation
print("\nTransformed DataFrame:\n", tips_df[['time', 'time_encoded', 'total_bill']].head())
```
