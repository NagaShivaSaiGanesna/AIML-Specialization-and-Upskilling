```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# 1. Create a sample dataset
df = pd.DataFrame({'color': ['red', 'blue', 'green', 'green', 'red']})

# 2. Initialize the encoder
label_encoder = LabelEncoder()

# 3. Fit and transform the data
# Note: LabelEncoder expects a 1D array/Series
df['encoded_color'] = label_encoder.fit_transform(df['color'])

print(df)
```

```python
# Transform a new data point
new_data = ['red']
print(label_encoder.transform(new_data)) # Output: [2]
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# 1. Create a sample dataset
df = pd.DataFrame({'size': ['small', 'medium', 'large', 'medium', 'small', 'large']})

# 2. Define the exact hierarchical order inside a list
# Note: It must be passed as a list of lists because OrdinalEncoder can handle multiple columns
size_order = [['small', 'medium', 'large']]

# 3. Initialize the encoder with the defined categories
ordinal_encoder = OrdinalEncoder(categories=size_order)

# 4. Fit and transform the data
# Note: OrdinalEncoder expects a 2D array/DataFrame
df['encoded_size'] = ordinal_encoder.fit_transform(df[['size']])

print(df)
```
