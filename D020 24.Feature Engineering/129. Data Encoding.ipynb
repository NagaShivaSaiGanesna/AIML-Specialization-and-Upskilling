```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Create a sample DataFrame
df = pd.DataFrame({'color': ['red', 'blue', 'blue', 'green', 'green', 'red', 'blue']})
print(df.head())
```

```python
# Create an instance of the OneHotEncoder
encoder = OneHotEncoder()

# Fit and transform the 'color' column. 
# Note: We pass df[['color']] to ensure it's a 2D array.
encoded_data = encoder.fit_transform(df[['color']])

# The output is a sparse matrix. Convert it to a dense array to view the values.
encoded_array = encoded_data.toarray()
print(encoded_array)
```

```python
# Get the new column names (e.g., 'color_blue', 'color_green', 'color_red')
column_names = encoder.get_feature_names_out()

# Create a new DataFrame with the encoded data
encoded_df = pd.DataFrame(encoded_array, columns=column_names)
print(encoded_df)
```

```python
# Concatenate horizontally (axis=1)
final_df = pd.concat([df, encoded_df], axis=1)
print(final_df)
```
