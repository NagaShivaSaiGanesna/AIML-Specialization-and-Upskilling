```python
import pandas as pd
import numpy as np
from sklearn.utils import resample

# Set seed for reproducibility
np.random.seed(42)

# Create 1000 data points
n_samples = 12
class_0_ratio = 0.9
n_class_0 = int(n_samples * class_0_ratio) # 900
n_class_1 = n_samples - n_class_0           # 100

# Generate Features (Normal Distribution)
class_0_data = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1_data = pd.DataFrame({
    'feature_1': np.random.normal(loc=2, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=2, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})

# Combine into one DataFrame
df = pd.concat([class_0_data, class_1_data]).reset_index(drop=True)

print(df['target'].value_counts())
# Output: 0: 900, 1: 100
```

```python
class_0_data
```

```python
df_majority = df[df.target == 0]
df_minority = df[df.target == 1]
```

```python
df_majority
```

## Demonstration: Upsampling

```python
# Upsample Minority Class
df_minority_upsampled = resample(df_minority, 
                                 replace=True,     # Sample with replacement
                                 n_samples=len(df_majority), # To match 900
                                 random_state=42)

# Combine back
df_upsampled = pd.concat([df_majority, df_minority_upsampled])

print(df_upsampled['target'].value_counts())
# Output: 0: 900, 1: 900
```

```python
df_minority_upsampled
```

## Demonstartion: Downsample

```python
# Downsample Majority Class
df_majority_downsampled = resample(df_majority, 
                                   replace=False,    # No replacement needed
                                   n_samples=len(df_minority), # To match 100
                                   random_state=42)

# Combine back
df_downsampled = pd.concat([df_minority, df_majority_downsampled])

print(df_downsampled['target'].value_counts())
# Output: 1: 100, 0: 100
```
