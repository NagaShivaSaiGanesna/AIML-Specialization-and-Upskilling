```python
from sklearn.datasets import make_classification
import pandas as pd

# Create imbalanced data: 1000 samples, 2 features, 90/10 weight
#by default the n_classes = 2
X, y = make_classification(n_samples=1000, n_features=2, n_redundant=0, 
                           n_clusters_per_class=1, weights=[0.90], random_state=12)

df = pd.DataFrame(X, columns=['f1', 'f2'])
df['target'] = y
print(df['target'].value_counts()) 
# Output: 0: 900, 1: 100
```

```python
# pip install imbalanced-learn
from imblearn.over_sampling import SMOTE

# Initialize SMOTE
oversample = SMOTE()

# Fit and transform the data
X_resampled, y_resampled = oversample.fit_resample(df[['f1', 'f2']], df['target'])

print(pd.Series(y_resampled).value_counts())
# Output: 0: 900, 1: 900
```
