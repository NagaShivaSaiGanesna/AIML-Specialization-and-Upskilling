```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_excel('flight_price.xlsx')
df.head()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.isnull().sum()
```

```python
df
```

```python
df['Date'] = df['Date_of_Journey'].str.split('/').str[0].astype(int)
```

```python
df['Month'] = df['Date_of_Journey'].str.split('/').str[1].astype(int)
```

```python
df['Year'] = df['Date_of_Journey'].str.split('/').str[2].astype(int)
```

```python
df.drop('Date_of_Journey',axis=1,inplace=True)
```

```python
df
```

```python
df['Arrival_Time'] =  df['Arrival_Time'].apply(lambda x: x.split(' ')[0])
```

```python
df
```

```python
# 1. Check and handle missing values (NaN)
print(df['Total_Stops'].isnull().sum()) 

# 2. Define the Ordinal Mapping Dictionary
stop_mapping = {
    'non-stop': 0,
    '1 stop': 1,
    '2 stops': 2,
    '3 stops': 3,
    '4 stops': 4,
    np.nan: 1  # Replacing NaN directly with the mode rank
}

# 3. Apply the mapping
df['Total_Stops'] = df['Total_Stops'].map(stop_mapping)
```

```python
from sklearn.preprocessing import OneHotEncoder

# Initialize the encoder
encoder = OneHotEncoder()

# Fit and transform the nominal columns
# You should encode all your nominal categorical variables at once.
encoded_features = encoder.fit_transform(df[['Airline', 'Source', 'Destination']])

# Convert the sparse matrix back into a readable DataFrame
encoded_df = pd.DataFrame(encoded_features.toarray(), columns=encoder.get_feature_names_out())

# Note: In a full pipeline, you would concatenate 'encoded_df' with 'df' and drop the original text columns.
```
